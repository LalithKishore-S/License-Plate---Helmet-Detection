```python
import json
import os
import cv2

def crop_head_regions(annotation_file, image_folder, output_folder_head, helmet_category_id=1):
    """
    Crops only the head regions (helmeted and estimated heads) and stores them separately.

    Args:
    - annotation_file: Path to the COCO annotation JSON containing helmet and estimated head annotations.
    - image_folder: Path to the folder containing original cropped two-wheeler images.
    - output_folder_head: Folder to save cropped head images.
    - helmet_category_id: COCO category ID for helmets.
    """

    # Create separate directories for helmet and no-helmet
    helmet_folder = os.path.join(output_folder_head, "helmet")
    no_helmet_folder = os.path.join(output_folder_head, "no_helmet")
    os.makedirs(helmet_folder, exist_ok=True)
    os.makedirs(no_helmet_folder, exist_ok=True)

    # Load COCO annotations
    with open(annotation_file, "r") as f:
        coco_data = json.load(f)

    # Load image info
    annotations = coco_data["annotations"]
    image_info = {img["id"]: img["file_name"] for img in coco_data["images"]}

    # Process each annotation (helmet or estimated head)
    for ann in annotations:
        img_id = ann["image_id"]
        bbox = ann["bbox"]  # [x, y, width, height]

        # Ensure the corresponding image exists
        if img_id not in image_info:
            print(f"Warning: No image found for ID {img_id}")
            continue

        img_path = os.path.join(image_folder, image_info[img_id])
        if not os.path.exists(img_path):
            print(f"Warning: Image {img_path} not found!")
            continue

        # Load the two-wheeler image
        img = cv2.imread(img_path)
        if img is None:
            print(f"Error: Unable to read image {img_path}")
            continue

        img_h, img_w, _ = img.shape
        x, y, w, h = map(int, bbox)

        # Ensure bounding box is within image dimensions
        if x < 0 or y < 0 or (x + w) > img_w or (y + h) > img_h:
            print(f"Warning: Bounding box {bbox} out of image bounds for {img_path}")
            continue

        # Crop the head region
        head_crop = img[y:y+h, x:x+w]

        # Ensure cropped image is not empty
        if head_crop.size == 0:
            print(f"Error: Empty cropped head for {img_path}, bbox: {bbox}")
            continue

        # Apply median blur (for slight noise reduction)
        head_crop = cv2.medianBlur(head_crop, 5)

        # Determine save path based on category
        if ann["category_id"] == helmet_category_id:
            folder = helmet_folder
            label = "helmet"
        else:
            folder = no_helmet_folder
            label = "no_helmet"

        head_filename = f"{img_id}_{label}.jpg"
        head_path = os.path.join(folder, head_filename)

        cv2.imwrite(head_path, head_crop)
        print(f"Saved: {head_path}")

    print(f"\nHead cropping complete! Images saved in:\n🔹 Helmet: '{helmet_folder}'\n🔹 No-Helmet: '{no_helmet_folder}'")
```

```python
annotation_file = "Data_helmet/train/_annotations.json"  # Update with actual path
image_folder = "Data_helmet/train"  # Folder containing cropped two-wheeler images
output_folder_head = "Data_helmet_detection_CNN/train"  # Folder to save head images

crop_head_regions(annotation_file, image_folder, output_folder_head)
```
