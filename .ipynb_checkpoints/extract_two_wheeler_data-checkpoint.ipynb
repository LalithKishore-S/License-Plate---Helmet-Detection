```python
import json
import shutil
import os

def filter_motorcycles(annotation_file, image_folder, output_annotation_file, output_image_folder, motorcycle_category_id=3):
    """
    Filters only motorcycle annotations from a COCO dataset and copies relevant images.
    
    Args:
    - annotation_file: Path to input COCO annotation JSON.
    - image_folder: Path to folder containing images.
    - output_annotation_file: Path to save filtered annotations.
    - output_image_folder: Path to save filtered images.
    - motorcycle_category_id: Category ID for motorcycles.
    """

    # Ensure output folder exists
    os.makedirs(output_image_folder, exist_ok=True)

    # Load COCO annotation file
    with open(annotation_file, "r") as f:
        coco_data = json.load(f)

    # Filter annotations for motorcycles
    filtered_annotations = [
        ann for ann in coco_data["annotations"] if ann["category_id"] == motorcycle_category_id
    ]

    # Get image IDs that contain motorcycles
    motorcycle_image_ids = {ann["image_id"] for ann in filtered_annotations}

    # Filter image entries
    filtered_images = [img for img in coco_data["images"] if img["id"] in motorcycle_image_ids]

    # Copy relevant images to the new output folder
    for img in filtered_images:
        src_path = os.path.join(image_folder, img["file_name"])
        dest_path = os.path.join(output_image_folder, img["file_name"])
        if os.path.exists(src_path):  # Avoid errors if some images are missing
            shutil.copy(src_path, dest_path)

    # Create new COCO JSON file with filtered data
    filtered_coco_data = {
        "images": filtered_images,
        "annotations": filtered_annotations,
        "categories": [cat for cat in coco_data["categories"] if cat["id"] == motorcycle_category_id]
    }

    # Save new annotations file
    with open(output_annotation_file, "w") as f:
        json.dump(filtered_coco_data, f, indent=4)

    print(f"Processed: {len(filtered_images)} images saved in '{output_image_folder}', annotations in '{output_annotation_file}'")
```

```python
filter_motorcycles(
    annotation_file="Data/train/_annotations.coco.json",
    image_folder="Data/train/",
    output_annotation_file="Data_twowheeler/train/_two_wheeler_annotations_train.json",
    output_image_folder="Data_twowheeler/train",
)
```

```python
filter_motorcycles(
    annotation_file="Data/test/_annotations.coco.json",
    image_folder="Data/test/",
    output_annotation_file="Data_twowheeler/test/_two_wheeler_annotations_test.json",
    output_image_folder="Data_twowheeler/test",
)
```
