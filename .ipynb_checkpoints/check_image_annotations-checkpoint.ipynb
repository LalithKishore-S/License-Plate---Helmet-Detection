Original Data

```python
import json
import cv2
import matplotlib.pyplot as plt

# Load COCO annotations file
with open("Data/train/_annotations.coco.json", "r") as f:
    data = json.load(f)

# Define category colors
category_colors = {
    1: (255, 0, 0),   # Helmet - Blue
    2: (0, 255, 0),   # License Plate - Green
    3: (255, 165, 0)  # Motorcyclist - Orange
}

# Select the first image
first_image_info = data["images"][0]  # Pick the first image
image_id = first_image_info["id"]
image_path = f"Data/train/{first_image_info['file_name']}"  # Change path!

# Load the image
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Get all annotations for the selected image
image_annotations = [ann for ann in data["annotations"] if ann["image_id"] == image_id ]

# Draw bounding boxes
for ann in image_annotations:
    x, y, w, h = map(int, ann["bbox"])  # Convert bbox to integer values
    category_id = ann["category_id"]
    color = category_colors.get(category_id, (255, 255, 255))  # Default White

    cv2.rectangle(image, (x, y), (x + w, y + h), color, 2)
    category_name = next(cat["name"] for cat in data["categories"] if cat["id"] == category_id)
    cv2.putText(image, category_name, (x, y - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

# Show the image
plt.figure(figsize=(8, 6))
plt.imshow(image)
plt.axis("off")
plt.title(f"Annotations for Image: {first_image_info['file_name']}")
plt.show()
```

License plate

```python
import json
import cv2
import matplotlib.pyplot as plt

# Load COCO annotations file
with open("Data_license/train/_annotations.json", "r") as f:
    data = json.load(f)

# Define category colors
category_colors = {
    1: (255, 0, 0),   # Helmet - Blue
    2: (0, 255, 0),   # License Plate - Green
    3: (255, 165, 0)  # Motorcyclist - Orange
}

# Select the first image
first_image_info = data["images"][1]  # Pick the first image
image_id = first_image_info["id"]
image_path = f"Data_license/train/{first_image_info['file_name']}"  # Change path!

# Load the image
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image = cv2.medianBlur(image,5)

# Get all annotations for the selected image
image_annotations = [ann for ann in data["annotations"] if ann["image_id"] == image_id ]

# Draw bounding boxes
for ann in image_annotations:
    x, y, w, h = map(int, ann["bbox"])  # Convert bbox to integer values
    category_id = ann["category_id"]
    color = category_colors.get(category_id, (255, 255, 255))  # Default White

    cv2.rectangle(image, (x, y), (x + w, y + h), color, 2)
    category_name = next(cat["name"] for cat in data["categories"] if cat["id"] == category_id)
    cv2.putText(image, category_name, (x, y - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

# Show the image
plt.figure(figsize=(8, 6))
plt.imshow(image)
plt.axis("off")
plt.title(f"Annotations for Image: {first_image_info['file_name']}")
plt.show()
```
