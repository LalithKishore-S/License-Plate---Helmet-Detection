```python
from ultralytics import YOLO
import torch
import os
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Using device: {device}")
model = YOLO(f'yolov8s.pt')
```

```python
data_yaml = "yolo_twowheeler_dataset/dataset.yaml"

model.train(
        data=data_yaml,
        epochs=100,
        imgsz=640,
        batch=16 ,  # Adjust batch size based on model size
        device=device,
        optimizer='adam',
        lr0=0.001,  # Initial learning rate
        name=f'twowheeler_yolov8',  # Experiment name
        patience=50,  # Early stopping patience
        save=True,
        save_period=10,  # Save checkpoint every 10 epochs
        single_cls=True,  # Set to True if you have only one class
)
```

```python
# Save trained model
model.save("models/twowheeler_model_yolov9_final.pt")
print("Model saved as 'twowheeler_model_yolov9_final.pt'")
```

```python
def evaluate_model(model, dataset_yaml):
    """Evaluate trained model on validation set"""
    metrics = model.val(
        data=dataset_yaml,
        split='val',  # or 'test'
        imgsz=640,
        conf=0.25,  # Confidence threshold
        iou=0.45,  # IoU threshold
        device='0' if torch.cuda.is_available() else 'cpu'
    )
    
    # Print key metrics
    print("\nEvaluation Results:")
    print(f"Precision: {metrics.box.map:.3f}")  # mAP@0.5
    print(f"Recall: {metrics.box.map50:.3f}")  # mAP@0.5:0.95
    print(f"mAP@0.5:0.95: {metrics.box.map75:.3f}")
    
    return metrics

# Run evaluation
val_metrics = evaluate_model(model, data_yaml)
```

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# Load your trained model
model = YOLO("models/twowheeler_model_yolov9_final.pt")

# Run inference on a single image
image_path = "Data_twowheeler/test/image_474_jpg.rf.16aabd29cb7afb05117d5fac449a9f22.jpg"
results = model.predict(source=image_path, conf=0.5, save=True)  # saves to runs/detect/predict

# Display with matplotlib (better for notebooks)
# Save and reload the image
output_path = "temp_result.jpg"
cv2.imwrite(output_path, results[0].plot())

# Display in notebook
plt.imshow(plt.imread(output_path))
plt.axis('off')
plt.show()
```
